The Song Success Prediction Model project aims to create a Machine Learning Model that can predict if a song will be a hit in the early or pre-release stages of production. The objective of this notebook is to train 3 Machine Learning Models and ultimately determine the best model to select as the final output of this project. In the steps below each model will be trained on the training data sets, then their accuracy will be tested using the test data sets. All three models will be objectively compared to make the final determination.

```python
def evaluate_classifier(model, X_test, y_test, model_name="Model"):
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)
    auc = roc_auc_score(y_test, y_prob)

    print(f"\n=== {model_name} ===")
    print(f"Accuracy : {acc:.4f}")
    print(f"Precision: {prec:.4f}")
    print(f"Recall   : {rec:.4f}")
    print(f"F1 Score : {f1:.4f}")
    print(f"ROC AUC  : {auc:.4f}")

    print("\nClassification Report:")
    print(classification_report(y_test, y_pred, zero_division=0))

    ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
    plt.title(f"Confusion Matrix - {model_name}")
    plt.show()

    return {
        "Model": model_name,
        "Accuracy": acc,
        "Precision": prec,
        "Recall": rec,
        "F1": f1,
        "ROC_AUC": auc
    }
```

# Model 1 Logistic Regression

```python
log_reg = LogisticRegression(
    max_iter=2000,
    class_weight="balanced"  # important due to imbalance
)

log_reg.fit(X_train_scaled, y_train)

results = []
results.append(evaluate_classifier(log_reg, X_test_scaled, y_test, "Logistic Regression"))
```

# Model 2 - Random Forest Classifier

```python
rf = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)

rf.fit(X_train_scaled, y_train)

results.append(evaluate_classifier(rf, X_test_scaled, y_test, "Random Forest"))
```

```python
y_prob_rf = rf.predict_proba(X_test_scaled)[:,1]

print("Min:", y_prob_rf.min())
print("Max:", y_prob_rf.max())
print("Mean:", y_prob_rf.mean())
```

```python
y_pred_rf = (y_prob_rf >= 0.15).astype(int)

from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred_rf))
```

# Model 3 - Gradient Boosting Classifier

```python
gb = GradientBoostingClassifier(
    n_estimators=200,
    learning_rate=0.05,
    random_state=42
)

gb.fit(X_train_scaled, y_train)

results.append(evaluate_classifier(gb, X_test_scaled, y_test, "Gradient Boosting"))
```

# Compare Models

```python
results_df = pd.DataFrame(results).sort_values(by="F1", ascending=False)
results_df
```

```python
results_df.set_index("Model")[["F1", "ROC_AUC", "Recall", "Precision"]].plot(
    kind="bar",
    figsize=(10, 5)
)
plt.title("Model Comparison Metrics")
plt.ylabel("Score")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))

for model, name in [(log_reg, "Logistic"), (rf, "Random Forest"), (gb, "Grad Boost")]:
    y_prob = model.predict_proba(X_test_scaled)[:,1]
    fpr, tpr, _ = roc_curve(y_test, y_prob)
    plt.plot(fpr, tpr, label=name)

plt.plot([0, 1], [0, 1], "--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curves")
plt.legend()
plt.show()
```

```python
final_model = results_df.iloc[0]["Model"]  # name only, we'll set model object below
best_model = rf if results_df.iloc[0]["Model"] == "Random Forest" else gb if results_df.iloc[0]["Model"] == "Gradient Boosting" else log_reg

y_prob = best_model.predict_proba(X_test_scaled)[:,1]
precision, recall, thresholds = precision_recall_curve(y_test, y_prob)

f1_scores = (2 * precision * recall) / (precision + recall + 1e-9)
best_idx = np.argmax(f1_scores)
best_threshold = thresholds[best_idx]

print("Best threshold for max F1:", best_threshold)
print("Best F1:", f1_scores[best_idx])
```

```python
y_pred_tuned = (y_prob >= best_threshold).astype(int)

print("Tuned Precision:", precision_score(y_test, y_pred_tuned, zero_division=0))
print("Tuned Recall   :", recall_score(y_test, y_pred_tuned, zero_division=0))
print("Tuned F1       :", f1_score(y_test, y_pred_tuned, zero_division=0))

ConfusionMatrixDisplay.from_predictions(y_test, y_pred_tuned)
plt.title("Confusion Matrix - Tuned Threshold")
plt.show()
```
